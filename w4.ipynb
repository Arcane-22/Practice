Create a bar chart for the car data above.

```python
import matplotlib.pyplot as plt

car_color = ['Red', 'Silver', 'Blue', 'White', 'Green']
values = [123, 230, 78, 193, 12]

plt.barh(car_color, values)
plt.title('Car seen')
plt.show()
```

Implement the code “ plt.barh(x, y)” to plot the bar chart, what is the effect?

```python
import matplotlib.pyplot as plt

car_color = ['Red', 'Silver', 'Blue', 'White', 'Green']
values = [123, 230, 78, 193, 12]
color = ['red', 'silver', 'blue', 'orange', 'green']

plt.barh(car_color, values, color = color)
plt.title('Car seen')
plt.show()
```

Create a list of colors and using the “ plt.bar(x, y, color = {you list here} ) “set the bar
chart colors to be the same as the car colors each bar represents

```python
myexplode = [0.3, 0.3, 0, 0, 0]
plt.pie(values, labels=car_color, explode=myexplode, shadow=True)
plt.title('Car Seen')
plt.show()
```
